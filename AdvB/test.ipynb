```python
import aenv
import EXP3
import numpy as np
```

```python
armA = "0"*750 + "1"*250
armB = "0"*250 + "1"*750

# expected 500 guessing randomly

rng = np.random.default_rng()
shuffledA = rng.permutation(list(armA))
shuffledB = rng.permutation(list(armB))

shuffledA = " ".join(shuffledA)
shuffledB = " ".join(shuffledB)

MAB = aenv.adversial_model(shuffledA, shuffledB)
exp3_pol = EXP3.EXP3_policy(MAB)

cumm_rew = [0]
arms = []
for t in range(1000):
    arm, rew = exp3_pol.step(time=t)
    arms.append(arm)
    cumm_rew.append(cumm_rew[t] + rew)

print(cumm_rew[-1])
print(arms)
```
