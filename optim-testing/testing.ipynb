```python
from bern_stochastic import model
from ucb1_optimised import ucb1_policy   
from utils import estimate_exponent
import matplotlib.pyplot as plt
import numpy as np

T = 100000
K = 16

ps_X = list(np.random.uniform(0, 1, K))
ps_A = [0.5] * K
pers = [1, int(np.sqrt(T)), T]

mmodel = model(K, T)
mmodel.psx = ps_X
mmodel.ps = ps_A
mmodel.pers = 10

mmodel.make_X()
mmodel.make_A()
```

```python
for x in pers:
    avgs = np.zeros(T)
    mmodel.pers = x
    for i in range(1, 61):
        mmodel.make_X()
        mmodel.make_A()
        policy = ucb1_policy(mmodel)
        ans = policy.play(weak=True)
        avgs += (ans - avgs) / i
    plt.plot(avgs, label=x)
    plt.legend()
plt.show()
```

```python
from exp3_optimised import exp3_policy

mmodel.make_X()
for x in pers:
    mmodel.pers = x
    mmodel.make_A()
    policy = exp3_policy(mmodel)
    ans = policy.play(weak=True)
    plt.plot(ans, label=x)
    plt.legend()

plt.show()
```
