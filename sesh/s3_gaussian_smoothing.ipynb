```python
from src.models.adversarial import bernoulli_iid_model
import numpy as np

np.random.seed(123)

K = 10
T = 1000

p = [0.5]*9 + [0.9]
s = [0.5]*10

rng = np.random.default_rng(12)

model = bernoulli_iid_model(K, T)
model.gen_bern(rng, p)
model.smooth_gaussian(s)
print(model.best)
```
